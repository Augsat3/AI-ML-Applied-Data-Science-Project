## Step 1 — Import libraries and load data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid', context='notebook')
DATA = Path('NSMES1988updated.csv')
FIG = Path('figures')
FIG.mkdir(exist_ok=True)

df = pd.read_csv(DATA)
print('shape:', df.shape)
df.head()
```

## Step 2 — Week 3 categorical plots: Health and Region

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.countplot(data=df, x='health', order=df['health'].value_counts().index, ax=axes[0], color='#4C72B0')
axes[0].set_title('Self-perceived Health')
axes[0].set_xlabel('health')
axes[0].set_ylabel('count')

sns.countplot(data=df, x='region', order=df['region'].value_counts().index, ax=axes[1], color='#55A868')
axes[1].set_title('Region')
axes[1].set_xlabel('region')
axes[1].set_ylabel('count')

plt.tight_layout()
plt.savefig(FIG / '01_health_region_counts.png', dpi=120, bbox_inches='tight')
plt.show()
```

```python
# Health x Region heatmap (counts) — matches Capstone 3 pivot
ct = pd.crosstab(df['health'], df['region'])
plt.figure(figsize=(8, 4))
sns.heatmap(ct, annot=True, fmt='d', cmap='Blues')
plt.title('Health x Region counts')
plt.tight_layout()
plt.savefig(FIG / '02_health_region_heatmap.png', dpi=120, bbox_inches='tight')
plt.show()
```

```python
# Mean physician visits by Health and by Region
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=df, x='health', y='visits', estimator='mean', errorbar=None, ax=axes[0], color='#C44E52')
axes[0].set_title('Mean visits by Health')
axes[0].set_ylabel('mean visits')

sns.barplot(data=df, x='region', y='visits', estimator='mean', errorbar=None, ax=axes[1], color='#8172B3')
axes[1].set_title('Mean visits by Region')
axes[1].set_ylabel('mean visits')

plt.tight_layout()
plt.savefig(FIG / '03_mean_visits_health_region.png', dpi=120, bbox_inches='tight')
plt.show()
```

## Step 3 — Relationship / correlation analyses (Week 3–4)

```python
# Numeric correlation heatmap
num = df.select_dtypes(include=[np.number])
corr = num.corr(numeric_only=True)

plt.figure(figsize=(10, 8))
sns.heatmap(corr, cmap='coolwarm', center=0, annot=False)
plt.title('Correlation heatmap (numeric columns)')
plt.tight_layout()
plt.savefig(FIG / '04_correlation_heatmap.png', dpi=120, bbox_inches='tight')
plt.show()

print('Top |corr| pairs with age / income / visits:')
focus = ['age', 'income', 'visits', 'chronic', 'school', 'hospital', 'emergency']
focus = [c for c in focus if c in corr.columns]
print(corr.loc[focus, focus].round(3))
```

```python
# Age vs Income scatter (sample for readability)
sample = df.sample(n=min(1500, len(df)), random_state=42)
plt.figure(figsize=(7, 5))
sns.scatterplot(data=sample, x='age', y='income', hue='gender', alpha=0.45, s=18)
plt.title('Age vs Income (sample), by gender')
plt.tight_layout()
plt.savefig(FIG / '05_age_vs_income_by_gender.png', dpi=120, bbox_inches='tight')
plt.show()
```

```python
# Income by gender + income by region (boxplots)
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(data=df, x='gender', y='income', ax=axes[0])
axes[0].set_title('Income by Gender')
sns.boxplot(data=df, x='region', y='income', ax=axes[1])
axes[1].set_title('Income by Region')
plt.tight_layout()
plt.savefig(FIG / '06_income_boxplots.png', dpi=120, bbox_inches='tight')
plt.show()
```

```python
# Health by Gender stacked counts + Age group x Gender
bins = [0, 30, 40, 50, 60, 70, 80, 120]
labels = ['<30', '30-39', '40-49', '50-59', '60-69', '70-79', '80+']
df = df.copy()
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=False)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
hg = pd.crosstab(df['health'], df['gender'])
hg.plot(kind='bar', stacked=True, ax=axes[0], color=['#4C72B0', '#DD8452'])
axes[0].set_title('Health by Gender')
axes[0].tick_params(axis='x', rotation=0)

ag = pd.crosstab(df['age_group'], df['gender'])
ag.plot(kind='bar', ax=axes[1], color=['#4C72B0', '#DD8452'])
axes[1].set_title('Age group by Gender')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig(FIG / '07_health_age_by_gender.png', dpi=120, bbox_inches='tight')
plt.show()
```

```python
# Chronic conditions vs health; visits vs chronic
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(data=df, x='health', y='chronic', ax=axes[0])
axes[0].set_title('Chronic conditions by Health')
sns.scatterplot(data=sample, x='chronic', y='visits', alpha=0.35, s=16, ax=axes[1])
axes[1].set_title('Visits vs Chronic conditions (sample)')
plt.tight_layout()
plt.savefig(FIG / '08_chronic_visits.png', dpi=120, bbox_inches='tight')
plt.show()
```

   The numeric heatmap summarizes linear relationships among age, income, schooling, visits, hospital/ER use, and chronic counts. Stronger cells are the first place to look for drivers Aura might use in demand or targeting features.

4. **Income relationships**  
   Age–income scatter (by gender) and income boxplots by gender/region visualize Capstone 3’s distribution tables: economic differences across demographics and geographies.

5. **Demographics**  
   Health-by-gender and age-group-by-gender charts make the Capstone 3 crosstabs readable at a glance for stakeholders.

6. **Aura takeaway**  
   Together these charts support the Aura story: aggregate healthcare-like person data, understand segments (health, region, gender, income), and surface relationships before modeling or recommendations.
